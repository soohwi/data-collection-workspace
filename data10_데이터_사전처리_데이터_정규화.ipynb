## 단위 환산 (Unit Conversion)
- 서로 다른 단위를 동일한 기준으로 통일
- 거리: km → m, 시간: 분 → 시간
- 단위가 다른 데이터를 비교 가능하게
- 단위 환산 이유: 어떤 계산이든 시간 단위가 통일되어야 오차가 없기 때문, 또한 외국 데이터인 경우  우리나라에 맞게 해줘 통일성을 주기 위해서

```python
import pandas as pd

# 데이터: 사람들의 이동 거리(km)와 시간(분)
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Distance_km': [5.2, 8.5, 3.1, 12.0],
    'Time_min': [15, 25, 10, 40]
}

df = pd.DataFrame(data)
print("원본 데이터:\n", df)

# 1. 단위 환산
# 거리: km → m (×1000), 시간: 분 → 시간 (÷60)
df['Distance_m'] = df['Distance_km'] * 1000
df['Time_hr'] = df['Time_min'] / 60

#  속도 : 거리(m) / 시간(초 or 시간)

print(df)
```

## 2. 자료형 변환


```python
import pandas as pd

# 예제 데이터
data = {
    'Age': ['25', '30', '40'],        # 문자열로 저장된 숫자
    'Salary': [50000.0, 60000.0, 70000.0],  # float
    'JoinDate': ['2023-01-01', '2023-02-15', '2023-03-10'],
    'Gender': ['Male', 'Female', 'Male']
}
df = pd.DataFrame(data)
print("기존 자료형:\n", df.dtypes)

# 숫자형 문자열 -> 정수형
df['Age'] = df['Age'].astype(int)  # 또는 'int64'

# 날짜 문자열 -> datetime 형식
df['JoinDate'] = pd.to_datetime(df['JoinDate'])

# 범주형 변환
df['Gender'] = df['Gender'].astype('category')

print("변환 후  자료형:\n", df.dtypes)
print(df)
```
