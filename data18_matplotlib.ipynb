# Matplotlib 시각화 

## 1 Matplotlib이란? (정의)

- **Matplotlib**은 파이썬에서 **2D 그래프 및 데이터 시각화를 작성하기 위해 만들어진 대표적인 저수준(Low-level) 오픈소스 라이브러리**입니다.
- MATLAB과 유사한 인터페이스를 제공하도록 설계되어, 과학 계산 및 데이터 분석 분야에서 가장 오랜 기간 널리 사용되고 있는 표준 시각화 도구입니다.

---

### 2 주요 특징 (Features)

- **세밀한 커스터마이징 (저수준 제어):** 그래프의 선 두께, 점 모양, 축 눈금, 범례 위치, 텍스트, 주석 등 모든 요소를 사용자가 직접 다룰 수 있습니다.
- **높은 자유도와 확장성:** 단순한 2D 차트부터 3D 차트, 애니메이션, 대화형(Interactive) 그래프까지 구현 가능합니다.
- **다른 라이브러리의 기반:** Pandas, Seaborn 등 파이썬의 주요 시각화 라이브러리들이 내부적으로 Matplotlib을 기반으로 구동됩니다.
- **다양한 파일 포맷 출력 지원:** PNG, JPG, SVG, PDF, EPS 등 고해상도 벡터/비트맵 서식을 모두 지원합니다.

---

### 3 Matplotlib의 객체 구조 (Figure & Axes)

Matplotlib을 제대로 이해하려면 핵심 구조인 **Figure**와 **Axes** 개념을 파악해야 합니다.

```
[Figure (전체 종이/캔버스)]
 ├── [Axes (하나의 그래프/차트)]
 │    ├── Title (제목)
 │    ├── X-axis / Y-axis (축)
 │    ├── Line / Scatter (데이터 표현 객체)
 │    └── Legend (범례)
 └── [Axes (또 다른 서브플롯 차트)]
```

- **Figure:** 전체 그래프가 들어가는 '도화지' 개념입니다.
- **Axes:** 실제 데이터가 그려지는 하나하나의 '그래프(축/서브플롯)' 영역입니다. 하나의 Figure 안에 여러 개의 Axes를 배치할 수 있습니다.

### 4 핵심 기능 및 차트 종류

**(1) Matplotlib 시각화 목적**
```
├── 1. 시간 흐름/추세 파악 ───► 선그래프 (plot), 면적 그래프 (stackplot), 계단형 그래프 (step)
├── 2. 항목 간 크기 비교 ────► 막대그래프 (bar/barh), 방사형/극좌표 차트 (polar)
├── 3. 변수 간 상관관계 ────► 산점도 (scatter), 벡터장/유선도 그래프 (quiver, streamplot)
├── 4. 분포 및 이상치 확인 ──► 히스토그램 (hist), 박스플롯 (boxplot), 바이올린플롯 (violinplot)
├── 5. 전체 대비 비율 확인 ──► 파이 차트 (pie), 넛/링 차트 (pie + wedgeprops)
└── 6. 3D & 행렬 시각화 ────► 3D 산점도/표면도 (scatter3D, plot_surface), 2D 이표 히트맵(imshow, matshow)
```

**(2) Matplotlib 주요 메서드**

| **목적** | **주요 메서드 / 그래프 종류** | **설명** |
| --- | --- | --- |
| **추세 파악** | `plt.plot()`, `plt.stackplot()`, `plt.step()` | 기본 선, 누적 면적 차트, 시계열 변화 단계 시각화 |
| **크기 비교** | `plt.bar()`, `plt.barh()`, `subplot(projection='polar')` | 수직/수평 막대 및 극좌표(Polar/Spider) 차트 |
| **상관 관계** | `plt.scatter()`, `plt.quiver()`, `plt.streamplot()` | 기본 2D 산점도 및 바람/유체 등의 방향 및 벡터장 시각화 |
| **분포/이상치** | `plt.hist()`, `plt.boxplot()`, `plt.violinplot()` | 도수분포표, 5수치 요약 상자그림 및 확률밀도 연계 바이올린 차트 |
| **비율 확인** | `plt.pie()` | 파이 차트 및 중앙 구멍(wedgeprops)을 통한 도넛 차트 구현 |
| **3D & 행렬** | `ax.plot_surface()`, `plt.imshow()`, `plt.matshow()` | 3차원 입체 표면도, 이미지/행렬 형태의 2D 데이터 출력 |
|  |  |  |

**(3) plot() 메서드 구성요소**

plot(*args: float | ArrayLike | str, scalex: bool = True, scaley: bool = True, data=None, **kwargs)

| 파라미터 | 설명 |
| --- | --- |
| `*args` | 보통 `x, y` 좌표값 또는 `y`값만 입력<br>예: `plot(x, y)`, 또는 `plot(y)` |
| `scalex` | `True`면 x축 자동 스케일 조정 (기본값 `True`) |
| `scaley` | `True`면 y축 자동 스케일 조정 (기본값 `True`) |
| `data` | `dict`, `DataFrame` 등으로 `x`, `y`를 이름으로 전달 가능<br>예: `data={'a':[1,2], 'b':[3,4]}` → `plot('a', 'b', data=data)` |
| `**kwargs` | 색상(color), 선 스타일(linestyle), 마커(marker) 등 추가 옵션 |

자주 사용하는 **kwargs (키워드 인자)

| 옵션 이름 | 예시 | 설명 |
| --- | --- | --- |
| `color` or `c` | `'blue'`, `'r'`, `'#FFAA00'` | 선 색상 |
| `linestyle` or `ls` | `'-'`, `'--'`, `':'`, `'-.'` | 선 스타일 |
| `linewidth` or `lw` | `1.5` | 선 굵기 |
| `marker` | `'o'`, `'s'`, `'x'`, `'^'` | 각 점에 표시할 마커 |
| `label` | `'실험값'` | 범례(legend) 이름 |
| `alpha` | `0.5` | 투명도 (0 \~ 1) |



### 5. 장점 및 단점

#### ⭕ 장점

1. **완벽한 제어권:** 원하고 상상하는 모든 모양의 그래프를 코드로 디테일하게 맞춤 설정(Customizing)할 수 있습니다.
2. **방대한 생태계 및 문서:** 역사가 오래되어 관련 예제 코드, 설명 문서, 스택오버플로우 Q&A가 가장 풍부합니다.
3. **높은 유연성:** 단순 EDA뿐만 아니라 논문 제출용, 학술용 고해상도 출판물 제작에 최적화되어 있습니다.

#### ❌ 단점

1. **많은 코드량:** 기본 그래프 설정에 써야 하는 코드 수가 판다스나 시본에 비해 상대적으로 길고 번거롭습니다.
2. **투박한 기본 디자인:** 추가 세팅을 하지 않으면 기본 그래프의 스타일이 세련되지 않을 수 있습니다.
3. **통계 기능 부재:** Seaborn처럼 회귀선이나 신뢰구간을 자동으로 계산해 주는 고수준 통계 시각화 기능은 직접 구현해야 합니다.




> **"파이썬 시각화의 든든한 뿌리이자, 세밀한 맞춤 설정이 가능한 만능 시각화 도구"**
> Pandas나 Seaborn으로 빠르게 그래프를 그린 뒤, **세부 레이아웃이나 축, 디자인을 완벽하게 다듬고 싶을 때** 반드시 다루게 되는 필수 라이브러리입니다.

# 2 Matplotlibe 그래프

## 2.1 데이터 준비

````python

from sklearn.datasets import load_diabetes

# 데이터 로드
diabetes = load_diabetes()
df = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
df['target'] = diabetes.target
````

## 2.2 선 그래프 (Line Plot)

````python
# bmi 값 순서대로 정렬 후 target 선형관계 확인
sorted_df = df.sort_values(by='bmi')
plt.figure(figsize=(8, 4))
plt.plot(sorted_df['bmi'].values, sorted_df['target'].values, color='blue', linestyle='--', marker='o')
plt.title('선 그래프: BMI vs Target')
plt.xlabel('BMI')
plt.ylabel('Target')
plt.grid(True)
plt.show()

print(type(sorted_df['bmi'])) #series

# 사용 데이터: bmi, target
# 용도: 시간 흐름, 순차적 변화, 연속적 관계 확인에 사용
# 해석: BMI가 증가할수록 Target(질병 진행 지표)도 증가하는 경향을 보임 → 양의 상관관계.
````

## 2.3 면적 그래프 (Area Plot)

````python
df_sorted = df.sort_values(by='bmi')
plt.figure(figsize=(8, 4))
plt.fill_between(df_sorted['bmi'], df_sorted['target'], color='lightgreen', alpha=0.5)
plt.title("면적 그래프: BMI vs Target")
plt.xlabel("BMI")
plt.ylabel("Target")
plt.grid(True)
plt.show()

# 사용 데이터: bmi, target
# 용도: 누적된 값, 면적 누적, 변화량 강조
# 해석: BMI 값이 커짐에 따라 Target이 함께 커지는 형태를 면적으로 표현하여 시각적으로 총 위험도 영역을 표현할 수 있음.
````

## 2.4 막대 그래프 (Bar Plot)

````python
# bmi를 세 그룹으로 나누고 평균 target 계산
df['bmi_group'] = pd.cut(df['bmi'], bins=3, labels=['Low', 'Medium', 'High'])
#print(df['bmi_group'])
grouped = df.groupby('bmi_group')['target'].mean()
print(grouped)
print(type(grouped))

plt.figure(figsize=(6, 4))
grouped.plot(kind='bar', color='pink')
plt.title("막대 그래프: BMI Group별 평균 Target")
plt.ylabel("평균 Target")
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.show()

# 사용 데이터: 범주형(bmi_group) vs 수치형(target 평균)
# 용도: 범주 간 비교에 적합
# 해석: BMI가 높을수록 평균 Target도 증가 → BMI가 높은 그룹일수록 질병 위험도 증가
````

## 2.5 히스토그램 (Histogram)

````python
plt.figure(figsize=(6, 4))
plt.hist(df['target'], bins=20, color='skyblue', edgecolor='black')
plt.title("히스토그램: Target 분포")
plt.xlabel("Target")
plt.ylabel("빈도")
plt.grid(True)
plt.show()

# 사용 데이터: 단일 연속형 변수 (target)
# 용도: 분포 형태 파악 (정규성, 치우침 등)
# 해석: 대략 정규분포 형태를 보이며, Target이 50~200 사이에 많이 분포함
````

## 2.6 산점도 (Scatter Plot)

````python
plt.figure(figsize=(6, 6))
plt.scatter(df['bmi'], df['target'], alpha=0.6)
plt.title("산점도: BMI vs Target")
plt.xlabel("BMI")
plt.ylabel("Target")
plt.grid(True)
plt.show()

# 사용 데이터: 두 수치형 변수
# 용도: 변수 간 상관관계 파악
# 해석: BMI가 높을수록 Target도 높아짐 → 선형 상관관계 있음
````

## 2.7 파이차트 (Pie Chart)

````python
# BMI 그룹별 인원 수
counts = df['bmi_group'].value_counts().sort_index()
print(counts)
plt.figure(figsize=(6, 6))
plt.pie(counts, labels=counts.index, autopct='%1.1f%%', colors=['lightblue', 'lightgreen', 'lightcoral'])
plt.title("파이차트: BMI Group 분포")
plt.show()

# 사용 데이터: 범주형 변수의 비율
# 용도: 비율 시각화, 그룹별 크기 비교
# 해석: 세 그룹(Low, Medium, High)이 대체로 균등하게 분포됨
````

## 2.8 박스플롯 (Box Plot)

````python
plt.figure(figsize=(6, 4))
plt.boxplot([df[df['bmi_group'] == g]['target'] for g in ['Low', 'Medium', 'High']],
            tick_labels=['Low', 'Medium', 'High'])
plt.title("박스플롯: BMI Group별 Target")
plt.ylabel("Target")
plt.grid(True)
plt.show()

# 사용 데이터: 그룹별 수치 분포
# 용도: 중앙값, 사분위, 이상치 확인
# 해석:
#   ① BMI가 높을수록 중앙값과 최대값 증가
#   ② High 그룹의 분포가 위로 치우쳐 있고, 위험 환자군 포함 가능성 높음
````
